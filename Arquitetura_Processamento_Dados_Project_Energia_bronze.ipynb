```python
# t - table
# his - history

# engg - energy generation
# comp - complaints
# engc - energy consume

# ing - ingestion
# csl - consolidated
# tgt - target
# ddm - data delivery modal, data mart

name_of_step = "csl"

engg_parquet_filename = "t_his_energy_generation_" + name_of_step + ".parquet"
comp_parquet_filename = "t_his_complaint_" + name_of_step + ".parquet"
engc_parquet_filename = "t_his_engc_" + name_of_step + ".parquet"
```
