```python
import requests
import pandas as pd
import numpy as np
#https://abve.org.br/bi-geral/
```

```python
from google.colab import drive
drive.mount('/content/drive')
google_drive_prefix = "/content/drive/MyDrive/cdai/csv/"
```

```python
# t - table
# his - history

# car_sells - energy generation

# ing - ingestion
# csl - consolidated
# tgt - target
# ddm - data delivery modal, data mart

name_of_step = "csl"

car_parquet_filename = "t_his_car_sells_" + name_of_step + ".parquet"

carparts_parquet_filename = "t_his_carparts_sells_" + name_of_step + ".parquet"
```

```python
df = pd.read_parquet(google_drive_prefix + car_parquet_filename, engine='pyarrow')

dfparts = pd.read_parquet(google_drive_prefix + carparts_parquet_filename, engine='pyarrow')
```

```python
dfparts
```

```python
df["Período"] = df["Período"].astype("category")

dfparts["Ano"] = dfparts["Ano"].astype("category")
```

```python
df['MHEV'] = df['MHEV'].astype(str)
df['MHEV'] = df['MHEV'].str.replace('.0', '')
```

```python
df['MHEV'] = df['MHEV'].str.replace('nan', '0')
```

```python
#df['MHEV'] = pd.to_numeric(df['MHEV'], errors="coerce")
df['MHEV'] = df['MHEV'].astype(np.int64)
```

```python
dfparts['Vendas'] = dfparts['Vendas'].astype(float)
```

```python
name_of_step = "tgt"

car_parquet_filename = "t_his_car_sells_" + name_of_step + ".parquet"

carparts_parquet_filename = "t_his_carparts_sells_" + name_of_step + ".parquet"


df.to_parquet(google_drive_prefix + car_parquet_filename, engine="pyarrow", compression="snappy")

dfparts.to_parquet(google_drive_prefix + carparts_parquet_filename, engine="pyarrow", compression="snappy")
```
